# 03 · geometry — α/β inter-domain docking angles

`pip install "kinapse[structures,geometry]"`

The ABangle-style **6 parameters** describe how the TCR's own α and β variable domains
sit relative to each other: **BA** (torsion), **BC1/BC2** & **AC1/AC2** (bend angles),
and **dc** (centroid distance). You get them per structure or **per MD frame**.

> Note: `example_ensemble.xtc` is a small *synthetic* ensemble generated from the example
> TCR, purely so this tutorial runs end-to-end. Swap in your own MD or generated ensemble.

```python
import os
os.environ.setdefault("ANARCI_CPU", "1")   # number on CPU (safe without a GPU)

from kinapse.structures import load_tcr

tcr = load_tcr("data/example_tcr.pdb", traj="data/example_ensemble.xtc", legacy_anarci=False)
pair = tcr.pairs[0]
print("frames:", pair.traj.n_frames)
```

## Per-frame geometry over the ensemble

```python
df = pair.calc_angles_traj()          # BA, BC1/AC1, BC2/AC2, dc — one row per frame
df.head()
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
ax = df.plot(x="frame", y=["BA"], legend=True, figsize=(6, 3))
ax.set_ylabel("BA torsion (deg)"); ax.set_title("α/β docking torsion across the ensemble")
plt.tight_layout()
```
