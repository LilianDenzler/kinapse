# 01 · Getting started

`kinapse` is a **lightweight core** plus **modular extras** — one concern per module,
installed only when you need it. This notebook: the module map, config, and loading
your first TCR.

**Install for this notebook**
```bash
pip install "kinapse[structures]"
```
Run this (and the others) from the `examples/` directory. TCR numbering uses the
pip-installable **ANARCII** via `legacy_anarci=False` (set `legacy_anarci=True` to use
bioconda ANARCI). The full module map is in [`../docs/MODULES.md`](../docs/MODULES.md).

````python
import kinapse
print("kinapse", kinapse.__version__)
````

## IMGT region constants (always available, no heavy deps)

````python
from kinapse import CDR_FR_RANGES, VARIABLE_RANGE
print("variable domain:", VARIABLE_RANGE)
{k: CDR_FR_RANGES[k] for k in ["A_CDR1", "A_CDR2", "A_CDR3", "B_CDR3"]}
````

## Config — paths from env/YAML, never hardcoded

````python
from kinapse.config import get_paths
paths = get_paths()
print("output dir:", paths.get_output_dir())
````

## Load your first TCR

````python
import os
os.environ.setdefault("ANARCI_CPU", "1")   # number on CPU (safe without a GPU)

from kinapse.structures import load_tcr

tcr = load_tcr("data/example_tcr.pdb", legacy_anarci=False)   # IMGT-renumber + pair α/β
pair = tcr.pairs[0]
print(len(tcr.pairs), "pair(s);  original chains:", pair.alpha_chain_id, "/", pair.beta_chain_id)
pair.cdr_fr_sequences()
````
