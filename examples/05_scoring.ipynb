# 05 · scoring — interface scoring of TCR-pMHC complexes

kinapse delegates interface scoring to the separate **ifscore** package (each scorer runs
in its own isolated env, so nothing bloats kinapse). See [`../docs/scoring.md`](../docs/scoring.md).

**Install**
```bash
pip install "ifscore @ git+https://github.com/LilianDenzler/scoring_functions"
ifscore install dockq prodigy      # provision scorers (uses uv); `geometry_scoring` needs nothing
```

````python
from kinapse import scoring
scoring.available_scorers()          # geometry_scoring, dockq, prodigy
````

## Score an interface
`geometry_scoring` (ifscore's interface-geometry scorer — BSA/SASA/contacts) needs no
external tool, so it runs immediately. Here we score the α–β interface of the example TCR
(receptor = chain A, ligand = chain B).

````python
scoring.score("data/example_tcr.pdb", rec=["A"], lig=["B"], scorers="geometry_scoring")
````

## TCR-pMHC complexes — auto-derive the chains
For a real **TCR-pMHC complex**, kinapse can fill in ifscore's receptor/ligand chains from
the loaded TCR (receptor = TCR α/β, ligand = pMHC = everything else):

```python
scoring.score_tcr_pmhc("data/complex.pdb", native="data/native.pdb",
                       scorers="default", legacy_anarci=False)
# or just the chains:
rec, lig = scoring.infer_tcr_pmhc_chains("data/complex.pdb", legacy_anarci=False)
```
(The example TCR here has no pMHC, so `score_tcr_pmhc` would raise — pass `rec`/`lig`
explicitly as above for a TCR-only interface.)

## From the command line
```bash
kinapse score complex.pdb --native native.pdb --auto-chains --new-anarci --scorers default
kinapse score --list-scorers
```
